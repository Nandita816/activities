```python
# -------------------------------
# INTRODUCTION TO PYTHON
# -------------------------------
# Python is a high-level, interpreted programming language
# used for web development, data science, AI, and automation.

# Example program
print("Hello Python")

# Output:
# Hello Python
```

```python
# -------------------------------
# DATA TYPES IN PYTHON
# -------------------------------
# Data types define the type of value stored in a variable.

# Integer (int) - whole numbers
x = 10
print("Integer:", x)  # Output: 10

# Float - decimal numbers
y = 10.5
print("Float:", y)  # Output: 10.5

# Boolean - True or False
z = 5 > 3
print("Boolean:", z)  # Output: True

# String - text data
name = "Python"
print("String:", name)  # Output: Python

# List - collection of values
nums = [1, 2, 3]
print("List:", nums)  # Output: [1, 2, 3]
```

```python
# -------------------------------
# VARIABLES
# -------------------------------
# Variable is a container used to store data values.

# Syntax: variable = value

a = 10
b = 20

print("a =", a)  # Output: 10
print("b =", b)  # Output: 20
```

```python
# -------------------------------
# EXPRESSIONS
# -------------------------------
# Expression is a combination of values and operators
# that produces a result.

x = 5 + 3 * 2

print("Result:", x)

# Output:
# Result: 11
```

```python
# -------------------------------
# STATEMENTS
# -------------------------------
# Statement is an instruction executed by Python interpreter.

print("This is a statement")

# Output:
# This is a statement
```

```python
# -------------------------------
# OPERATOR PRECEDENCE
# -------------------------------
# Python follows order:
# () > ** > * / > + -

x = 2 + 3 * 4
print("Result:", x)

# Output:
# Result: 14
```

```python
# -------------------------------
# COMMENTS
# -------------------------------
# Comments are ignored by Python and used for explanation.

# This is a single-line comment
print("Comments are useful")

# Output:
# Comments are useful
```

```python
# -------------------------------
# MODULES
# -------------------------------
# Module is a file containing reusable Python code.

import math

print("Square root:", math.sqrt(16))

# Output:
# Square root: 4.0
```

```python
# -------------------------------
# FUNCTIONS
# -------------------------------
# Function is a block of reusable code.

# Syntax:
# def function_name():
#     code

def greet():
    print("Hello")

greet()

# Output:
# Hello
```

```python
# -------------------------------
# FLOW OF EXECUTION
# -------------------------------
# Python executes code line by line from top to bottom.

print("First")
print("Second")

# Output:
# First
# Second
```

```python
# -------------------------------
# PARAMETERS & ARGUMENTS
# -------------------------------
# Parameters = variables in function definition
# Arguments = actual values passed

def add(a, b):
    return a + b

print("Sum:", add(2, 3))

# Output:
# Sum: 5
```

```python
# -------------------------------
# BOOLEAN OPERATORS
# -------------------------------
# Used for logical operations:
# and, or, not

x = 5 > 3 and 2 < 4
print("Result:", x)

# Output:
# True
```

```python
# -------------------------------
# IF STATEMENT
# -------------------------------
# Executes block only if condition is True.

x = 10

if x > 5:
    print("Greater")

# Output:
# Greater
```

```python
# -------------------------------
# IF-ELSE STATEMENT
# -------------------------------
# Chooses between two blocks.

x = 3

if x > 5:
    print("Yes")
else:
    print("No")

# Output:
# No
```

```python
# -------------------------------
# IF-ELIF-ELSE
# -------------------------------
# Used for multiple conditions.

x = 7

if x > 10:
    print("Big")
elif x > 5:
    print("Medium")
else:
    print("Small")

# Output:
# Medium
```

```python
# -------------------------------
# WHILE LOOP
# -------------------------------
# Repeats until condition becomes False.

i = 1

while i <= 3:
    print(i)
    i += 1

# Output:
# 1
# 2
# 3
```

```python
# -------------------------------
# FOR LOOP
# -------------------------------
# Used to iterate over sequence.

for i in range(3):
    print(i)

# Output:
# 0
# 1
# 2
```

```python
# -------------------------------
# BREAK
# -------------------------------
# Stops loop immediately.

for i in range(5):
    if i == 3:
        break
    print(i)

# Output:
# 0
# 1
# 2
```

```python
# -------------------------------
# CONTINUE
# -------------------------------
# Skips current iteration.

for i in range(5):
    if i == 3:
        continue
    print(i)

# Output:
# 0
# 1
# 2
# 4
```

```python
# -------------------------------
# FRUITFUL FUNCTIONS
# -------------------------------
# Function that returns a value.

def square(x):
    return x * x

print(square(4))

# Output:
# 16
```

```python
# -------------------------------
# SCOPE (LOCAL & GLOBAL)
# -------------------------------
# Local variable: inside function
# Global variable: outside function

x = 10

def func():
    x = 5
    print("Local:", x)

func()
print("Global:", x)

# Output:
# Local: 5
# Global: 10
```

```python
# -------------------------------
# FUNCTION COMPOSITION
# -------------------------------
# Using output of one function inside another.

def add(x):
    return x + 2

def mul(x):
    return x * 3

print(add(mul(2)))

# Output:
# 8
```

```python
# -------------------------------
# RECURSION
# -------------------------------
# Function calling itself.

def fact(n):
    if n == 1:
        return 1
    return n * fact(n - 1)

print(fact(5))

# Output:
# 120
```

```python
# -------------------------------
# STRING SLICING
# -------------------------------
# Extract part of string.

text = "Python"

print(text[0:3])

# Output:
# Pyt
```

```python
# -------------------------------
# IMMUTABILITY
# -------------------------------
# Strings cannot be changed after creation.

text = "Hi"
print(text)

# text[0] = "h"  # Error

# Output:
# Hi
```

```python
# -------------------------------
# STRING METHODS
# -------------------------------
# Built-in functions for string operations.

name = "python"

print(name.upper())

# Output:
# PYTHON
```

```python
# -------------------------------
# ARRAYS
# -------------------------------
# Stores same type of elements.

import array as arr

a = arr.array('i', [1, 2, 3])

print(a)

# Output:
# array('i', [1, 2, 3])
```

```python
# Access elements using index

print(a[0])

# Output:
# 1
```

```python
# Array methods like append

a.append(4)

print(a)

# Output:
# array('i', [1, 2, 3, 4])
```

```python
# list Ordered, changeable collection

a = [1, 2, 3]

print(a)

# Output:
# [1, 2, 3]
```

```python
# tuple Ordered but immutable

t = (1, 2, 3)

print(t)

# Output:
# (1, 2, 3)
```

```python
# dictionary Key-value pairs

d = {"name": "Python", "age": 30}

print(d)

# Output:
# {'name': 'Python', 'age': 30}
```
