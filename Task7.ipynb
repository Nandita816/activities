```python
LISTS IN PYTHON

1. What is a List?

A list is an ordered, mutable (changeable) collection in Python that can store multiple values of different data types.
Key Points:
Ordered (maintains insertion order)
Mutable (can be changed)
Allows duplicate values
Indexed (starts from 0)

Important Features of Lists:

- Can store multiple data types
- Can be modified after creation
- Supports indexing and slicing
- Supports duplicate values
```

```python
#2. Creating a List

# Empty list
a = []

# List with elements
fruits = ["Apple", "Banana", "Orange"]

print(fruits)
```

```python
#4. List Methods (Important)
#append()
#Adds an item at the end of the list.

fruits = ["Apple", "Banana"]
fruits.append("Mango")

print(fruits)
```

```python
#insert()
#Adds an item at a specific position.

fruits = ["Apple", "Banana"]
fruits.insert(1, "Grapes")

print(fruits)
```

```python
#remove()
#Removes a specific value.

fruits = ["Apple", "Banana", "Mango"]
fruits.remove("Banana")

print(fruits)
```

```python
#pop()
#Removes item using index (default last item).

fruits = ["Apple", "Banana", "Mango"]
fruits.pop()

print(fruits)
```

```python
#sort()
#Sorts list in ascending order.

numbers = [5, 2, 9, 1]
numbers.sort()

print(numbers)
```

```python
#reverse()
#Reverses the list order.

numbers = [1, 2, 3, 4]
numbers.reverse()

print(numbers)
```

```python
#count()
#Counts occurrences of a value.

numbers = [1, 2, 2, 3, 2]

print(numbers.count(2))
```

```python
#index()
#Returns index of first occurrence.

numbers = [10, 20, 30]

print(numbers.index(20))
```

```python
#extend()
#Adds multiple elements to list.

a = [1, 2]
b = [3, 4]

a.extend(b)

print(a)
```

```python
#clear()
#Removes all elements.

fruits = ["Apple", "Banana"]
fruits.clear()

print(fruits)
```

```python
#5. Useful Built-in Functions (with lists)

numbers = [10, 20, 30, 40]

print(len(numbers))   # length
print(max(numbers))   # maximum
print(min(numbers))   # minimum
print(sum(numbers))   # sum
```

```python
#6. List Slicing

numbers = [10, 20, 30, 40, 50]

print(numbers[1:4])   # slice
print(numbers[:3])    # start to index 3
print(numbers[::2])   # step slicing
```

```python
#7. Looping through a List

fruits = ["Apple", "Banana", "Mango"]

for item in fruits:
    print(item)
```

```python
TUPLES IN PYTHON 

1. What is a Tuple?
A tuple is an ordered collection in Python that is immutable (cannot be changed after creation).
    
Key Points:
    
Ordered (keeps insertion order)
Immutable (cannot modify after creation)
Allows duplicate values
Faster than lists (because it is fixed)
Written using round brackets ()

Important Features of Tuples:

- Immutable (cannot change values)
- Faster than lists
- Used for fixed data
- Supports indexing and slicing
```

```python
#2. Creating a Tuple

# Empty tuple
t1 = ()

# Tuple with elements
numbers = (10, 20, 30, 40)

print(numbers)
```

```python
#3. Accessing Elements

numbers = (10, 20, 30, 40)

print(numbers[0])   # First element
print(numbers[2])   # Third element
print(numbers[-1])  # Last element
```

```python
#4. Tuple Methods (Only 2 Methods)
#count()
#Returns how many times a value appears in the tuple.

numbers = (10, 20, 30, 20, 40, 20)

print(numbers.count(20))
```

```python
#5.index()
#Returns the index (position) of the first occurrence of a value.

numbers = (10, 20, 30, 40)

print(numbers.index(30))

#Why only few methods?

#Tuples are immutable (cannot be changed),
#so methods like append(), remove(), pop() are NOT available.
```

```python
#6.Built-in Functions used with Tuples

numbers = (10, 20, 30, 40)

print(len(numbers))   # number of elements
print(max(numbers))   # largest value
print(min(numbers))   # smallest value
print(sum(numbers))   # total sum
```

```python
#7. Tuple Operations
#7.1 Concatenation

a = (1, 2)
b = (3, 4)

print(a + b)
```

```python
#7.2 Repetition

t = (1, 2)

print(t * 3)
```

```python
#7.3 Membership

numbers = (10, 20, 30)

print(20 in numbers)
print(50 in numbers)
```

```python
#8. Tuple Slicing

numbers = (10, 20, 30, 40, 50)

print(numbers[1:4])   # slice
print(numbers[:3])    # first 3 elements
print(numbers[::2])   # step slicing
```

```python
#9.Looping Through Tuple

fruits = ("Apple", "Banana", "Mango")

for item in fruits:
    print(item)
```

```python
DICTIONARIES IN PYTHON 

1. What is a Dictionary?
A dictionary is a collection of data stored in key-value pairs. Dictionaries are mutable (changeable), ordered, and do not allow duplicate keys.

Key Points:

Stores data as key : value
Mutable (can be modified)
Ordered (Python 3.7+)
Keys must be unique
Values can be duplicated
Created using curly braces {}

Important Features of Dictionaries:

- Stores data as key-value pairs
- Mutable (can be modified)
- Keys must be unique
- Values can be duplicated
- Fast data retrieval using keys
- Supports various built-in methods

#2. Creating a Dictionary

# Empty dictionary
student = {}

# Dictionary with values
student = {
    "name": "John",
    "age": 21,
    "course": "Python"
}

print(student)

#3. Accessing Dictionary Values

student = {
    "name": "John",
    "age": 21,
    "course": "Python"
}

print(student["name"])
print(student["age"])


#4. Dictionary Methods
#4.1 get()
#Returns the value of a specified key.

student = {
    "name": "John",
    "age": 21
}

print(student.get("name"))

#4.2 keys()
#Returns all keys in the dictionary.

student = {
    "name": "John",
    "age": 21
}

print(student.keys())

#4.3 values()
#Returns all values in the dictionary.

student = {
    "name": "John",
    "age": 21
}

print(student.values())


#4.4 items()
#Returns all key-value pairs.

student = {
    "name": "John",
    "age": 21
}

print(student.items())

#4.5 update()
#Updates an existing value or adds a new key-value pair.

student = {
    "name": "John",
    "age": 21
}

student.update({"age": 22})

print(student)

#4.6 pop()
#Removes a specified key.

student = {
    "name": "John",
    "age": 21
}

student.pop("age")

print(student)

#4.7 popitem()
#Removes the last inserted key-value pair.

student = {
    "name": "John",
    "age": 21
}

student.popitem()

print(student)

#4.8 clear()
#Removes all items from the dictionary.
student = {
    "name": "John",
    "age": 21
}

student.clear()

print(student)

#4.9 copy()
#Creates a copy of the dictionary.
student = {
    "name": "John",
    "age": 21
}

new_student = student.copy()

print(new_student)

#5. Adding New Elements

student = {
    "name": "John"
}

student["age"] = 21

print(student)

#6. Built-in Functions with Dictionaries

student = {
    "name": "John",
    "age": 21,
    "course": "Python"
}

print(len(student))

#7. Looping Through a Dictionary
#Loop Through Keys

student = {
    "name": "John",
    "age": 21
}

for key in student:
    print(key)

#Loop Through Values
Python
student = {
    "name": "John",
    "age": 21
}

for value in student.values():
    print(value)

#Loop Through Key-Value Pairs

student = {
    "name": "John",
    "age": 21
}

for key, value in student.items():
    print(key, ":", value)

```
